```python
#for loop demo
a_list = [6.49, 2.49, 8.98, 2.0, 0.4900000000000002, 20.160100000000003]
new_list=[]
print(new_list)
for obj in a_list:
    print(obj)
    new_list.append(2*obj)
    print(new_list)
print(a_list)
```

```python
#Conditional Statements
#comparison
x = 3 #assignment operation 
if x == 5:
    print("Value is equal to 5")
#else if = elif 
elif x < 5: 
    print("Value is less than 5")
else:
    print("Value is greater than 5")
```

```python
#combining for loops and if else conditional statements 
num_list = [2,3,47,9,4,7,8,10,1,6,5]
y = 5
for num in num_list:
    if num == y:
        print(f"{num} is equal to {y}")
    elif num < y: 
        print(f":{num} is less than {y}")
    else: 
        print(f"{num} is greater than {y}")
              
```
